```python
import os
import json
import pickle
import joblib
import numpy as np
import pprint

def run_diagnostics():
    print("====== 🔍 ACE System Artifacts Diagnostics (Fixed) 🔍 ======\n")
    files = os.listdir('.')
    print(f"Detected files: {files}\n")
    
    report = {}
    
    # 1. فحص Label Encoder
    le_file = 'label_encoder_final (1).pkl'
    if le_file in files:
        try:
            le = joblib.load(le_file)
            report['label_encoder_classes'] = list(le.classes_)
        except Exception as e:
            report['label_encoder_error'] = str(e)
            
    # 2. فحص هيكل الحزمة stage1_production_bundle.pkl
    bundle_file = 'stage1_production_bundle.pkl'
    if bundle_file in files:
        try:
            bundle = joblib.load(bundle_file)
            report['bundle_type'] = str(type(bundle))
            if isinstance(bundle, dict):
                report['bundle_keys'] = list(bundle.keys())
                for k, v in bundle.items():
                    report[f'bundle_key_{k}_type'] = str(type(v))
            else:
                report['bundle_attributes'] = dir(bundle)[:15]
        except Exception as e:
            try:
                with open(bundle_file, 'rb') as f:
                    bundle = pickle.load(f)
                report['bundle_type_pickle'] = str(type(bundle))
                if isinstance(bundle, dict):
                    report['bundle_keys_pickle'] = list(bundle.keys())
            except Exception as e2:
                report['bundle_error'] = f"joblib: {e} | pickle: {e2}"

    # 3. فحص النموذج final_model.pkl
    model_file = 'final_model.pkl'
    if model_file in files:
        try:
            model = joblib.load(model_file)
            report['model_type'] = str(type(model))
            if 'Booster' in str(type(model)):
                report['model_features_count'] = len(model.feature_name())
                report['model_features_sample'] = model.feature_name()[:5]
            else:
                if hasattr(model, 'feature_name_'):
                    report['model_features_count'] = len(model.feature_name_)
                    report['model_features_sample'] = list(model.feature_name_)[:5]
                if hasattr(model, 'classes_'):
                    report['model_classes'] = [x.item() if hasattr(x, 'item') else x for x in model.classes_]
        except Exception as e:
            report['model_error'] = str(e)

    # 4. فحص المعالج preprocessing_pipeline_final.pkl
    pipe_file = 'preprocessing_pipeline_final.pkl'
    if pipe_file in files:
        try:
            pipe = joblib.load(pipe_file)
            report['pipeline_type'] = str(type(pipe))
            if hasattr(pipe, 'get_feature_names_out'):
                report['pipeline_out_features_sample'] = list(pipe.get_feature_names_out())[:5]
            elif hasattr(pipe, 'steps'):
                report['pipeline_steps'] = [step[0] for step in pipe.steps]
        except Exception as e:
            report['pipeline_error'] = str(e)

    # طباعة التقرير النهائي المنظم باستخدام pprint لمنع مشاكل الـ JSON
    pp = pprint.PrettyPrinter(indent=2)
    pp.pprint(report)
    print("\n==================================================")

if __name__ == "__main__":
    run_diagnostics()
```
